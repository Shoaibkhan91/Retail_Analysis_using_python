### 7. Which store type sells the maximum products by value and by quantity?

```python
h=cust_tran_pro.loc[cust_tran_pro.Qty >0]
most_sales_quan=h.groupby('Store_type')[['Qty','total_amt']].agg('sum')

most_sales_quan.sort_values('Qty',ascending=False).reset_index()
most_sales_quan.round(0)
most_sales_quan.head(1)

```

### 8. What was the total amount earned from the "Electronics" and "Clothing" categories from Flagship Stores

```python
h=cust_tran_pro.loc[
    (cust_tran_pro.Store_type=='Flagship store')  &
    (cust_tran_pro.prod_cat.isin(['Electronics','Clothing'])) & (cust_tran_pro.Qty>0)
    ]
g=h.groupby('prod_cat').total_amt.sum().reset_index()
total_earn=h['total_amt'].sum()
print('The Total Amount earned from the "Electronics" and "Clothing" categories from Flagship Stores : ',total_earn)
```

### 9. What was the total amount earned from "Male" customers under the "Electronics" category


```python
male_elect_earn=cust_tran_pro.loc[
    (cust_tran_pro.Gender=='M') &  (cust_tran_pro.prod_cat=='Electronics') &(cust_tran_pro.Qty > 0)
    ]
h=male_elect_earn.groupby('prod_cat').total_amt.sum().reset_index()
h.total_amt[0]
```

### 10. How many customers have more than 10 unique transactions, after removing all transactions which have any negative amounts?

```python
h=cust_tran_pro.loc[cust_tran_pro.Qty > 0]
p=h.groupby('customer_Id').transaction_id.nunique().reset_index()
p.loc[p.transaction_id > 10].shape[0]
```

### 11. For all customers aged between 25 - 35, find out:
### a. What was the total amount spent for “Electronics” and “Books” product categories?
 ### b. What was the total amount spent by these customers between 1st Jan, 2014 to 1st Mar, 2014?

### A

```python
today = pd.to_datetime('today')
cust_tran_pro['DOB']=pd.to_datetime(cust_tran_pro['DOB'],format='%d-%m-%Y')
```

```python
today = pd.to_datetime('today')
cust_tran_pro['Age'] = (today - cust_tran_pro['DOB']).dt.days // 365
mid_age=cust_tran_pro.loc[(cust_tran_pro.Age >=25) & (cust_tran_pro.Age <=35) & (cust_tran_pro.Qty >0)]
mid_age.loc[(mid_age.prod_cat=='Electronics') |  (mid_age.prod_cat=='Books') ].total_amt.sum()
```

### B

```python
cust_tran_pro['tran_date']=pd.to_datetime(cust_tran_pro['tran_date'],format='mixed')
mid_ages=cust_tran_pro.loc[(cust_tran_pro.Age >=25) & (cust_tran_pro.Age <=35) & (cust_tran_pro.Qty >0)]
mid_ages.loc[(mid_ages.tran_date >='01-01-2014' ) & (mid_ages.tran_date <='01-03-2014')].total_amt.sum()
```
