### 5. Analyze which product categories are more popular among females vs male customers

```python
popular_categories=cust_tran_prod.pivot_table(
    index='prod_cat',columns='Gender',values='transaction_id',aggfunc='count').reset_index().sort_values('F',ascending=False,ignore_index=True)
popular_categories.loc[::,'prod_cat'].head(1)
```

### 6. Which City code has the maximum customers and what was the percentage of customers from that city

```python
max_cust=cust_tran_prod.groupby('city_code').customer_Id.nunique().reset_index()
p=max_cust.sort_values('customer_Id',ascending=False,ignore_index=True)
p['customer_percent']=(p.customer_Id/p.customer_Id.sum() )*100
p.rename({'customer_Id':'No_of_customer'},axis=1,inplace=True)
top_city=p.head(1)
top_city.iloc[0]
```

 ### 7. Which store type sells the maximum products by value and by quantity

```python
max_product=cust_tran_prod.loc[(cust_tran_prod.Qty >  0 ) & (cust_tran_prod.total_amt > 0) ]
store_max=max_product.groupby('Store_type')[['total_amt','Qty']].sum().round({'total_amt':2}).sort_values('total_amt',ascending=False)
store_max.head(1)
```

### 8. What was the total amount earned from the "Electronics" and "Clothing" categories from Flagship Stores

```python
Total_amount=cust_tran_prod.loc[(cust_tran_prod.prod_cat.isin(['Electronics','Clothing']) ) &  (cust_tran_prod.Store_type == 'Flagship store') & (cust_tran_prod.Qty > 0) & (cust_tran_prod.total_amt > 0 )]
Total_amount.total_amt.sum()
```

### 9. What was the total amount earned from "Male" customers under the "Electronics" category?

```python
total_revenu_by_male=cust_tran_prod.loc[
    (cust_tran_prod.Gender == 'M') & (cust_tran_prod.prod_cat == 'Electronics') & (cust_tran_prod.total_amt > 0) & (cust_tran_prod.Qty > 0)]
total_revenu_by_male.total_amt.sum()
```

### 10. How many customers have more than 10 unique transactions, after removing all transactions which have any negative amounts?

```python
y=cust_tran_prod.loc[cust_tran_prod.total_amt > 0]
more_10=y.groupby('customer_Id').transaction_id.nunique().reset_index()
more_10.rename({'transaction_id':'No_of_nunique'},axis=1,inplace=True)
result=more_10.loc[more_10.No_of_nunique > 10].reset_index(drop=True)
result.shape[0]

```

#### 11. For all customers aged between 25 - 35, find out:
------------------------------------------------------------
#### a. What was the total amount spent for “Electronics” and “Books” product categories?


```python
spend=cust_tran_prod.loc[
    (cust_tran_prod.Age >=25) & (cust_tran_prod.Age <=35 ) & (cust_tran_prod.prod_cat.isin(['Electronics','Books'])) &(cust_tran_prod.total_amt > 0)]
spend.groupby('prod_cat').total_amt.sum().reset_index()
```

### b. What was the total amount spent by these customers between 1st Jan, 2014 to 1st Mar, 2014?

```python
spent=cust_tran_prod.loc[
    (cust_tran_prod.Age >=25)&(cust_tran_prod.Age <=35) & (cust_tran_prod.tran_date >='2014-01-01') & (cust_tran_prod.tran_date <= '2014-03-01') &(cust_tran_prod.total_amt > 0)]
spent.total_amt.sum()
```
